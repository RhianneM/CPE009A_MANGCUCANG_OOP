###Distance Formula

```python
from math import sqrt
coordinates_list = [(1, 1), (2, 3)]
p1 = coordinates_list[0]
p2 = coordinates_list[1]
distance = sqrt((p2[0] - p1[0])**2 + (p2[1] - p1[1])**2)
print(f"The distance between {p1} and {p2} is: {distance}")
```

###Simple World Filter

```python
text = input("Enter a string: ")
word_to_filter = "stupid"
replacement = "*" * len(word_to_filter)
filtered_text = text.replace(word_to_filter, replacement)
print("Filtered Text:", filtered_text)
```

###Phonebook

```python
phonebook = [
    {'id': 1, 'name': 'John Doe', 'contact': '09060611233', 'address': '123 Main St'},
    {'id': 2, 'name': 'Charles Leclerc', 'contact': '09171234567', 'address': '456 Oak Ave'},
    {'id': 3, 'name': 'Daniella Rosales', 'contact': '09187654321', 'address': '789 Pine Rd'}
]
try:  # ← Add this line
    search_id = int(input("Enter the ID of the record to find: "))
    found = False

    for person in phonebook:
        if person['id'] == search_id:
            print("\n--- Record Found ---")
            print(f"Name: {person['name']}")
            print(f"Contact: {person['contact']}")
            print(f"Address: {person['address']}")
            found = True
            break

    if not found:
        print(f"No record found with ID: {search_id}")

except ValueError:  # ← Now this works because it has a matching try
    print("Invalid input. Please enter a number for the ID.")
```
