**Procedure:**
1. A teacher wants to calculate the final grade in a CpE course and want to write it in a python program. The following are the requirements:
1. PRELIM GRADE = 50% Prelim Exam + 50% Prelim Class Standing (CS)
2. PRELIM CS = 50% Hands-on activity + 30% Quiz + 20% Assignment
3. MIDTERM GRADE = 1/3 of PRELIM GRADE + 2/3 of (50% Midterm Exam + 50% Midterm Class Standing (CS))
4. MIDTERM CS = 50% Hands-on activity + 30% Quiz + 20% Assignment
5. FINAL GRADE = 1/3 of MIDTERM GRADE + 2/3 of (50% Final Exam + 50% Final Class Standing (CS))
6. FINAL CS = 50% Hands-on activity + 30% Quiz + 20% Assignment
7. HOAs, Quizzes and Assignments are inputted as average of all submissions and are out of 100%.
8. Major exams are inputted out of 100%.
9. Show the codes that successfully run the program.
10. Provide comments or documentation strings for your program.

```python
# Place all the output of the procedures here
PRELIMHOA=float(input("\nEnter Hands-on Activity grade: "))
PRELIMQUIZ=float(input("Enter Prelim Quiz grade: "))
PRELIMAS=float(input("Enter Prelim Assignment grade: "))
PRELIMCS=PRELIMHOA*.5+PRELIMQUIZ*.3+PRELIMAS*.2
PRELIMEX=float(input("Enter Prelim Exam grade: "))
PRELIMG=PRELIMEX*.5+PRELIMCS*.5
print("Prelim Grades is: ", "%.2f" %PRELIMG)

MIDTERMHOA=float(input("\nEnter Hands-on Activity grade: "))
MIDTERMQUIZ=float(input("Enter Midterm Quiz grade: "))
MIDTERMAS=float(input("Enter Midterm Assignment grade: "))
MIDTERMCS=MIDTERMHOA*.5+MIDTERMQUIZ*.3+MIDTERMAS*.2
print("Midterm CS is: ", MIDTERMCS)
MIDTERMEX=float(input("Enter Midterm Exam grade: "))
MIDTERMG=MIDTERMEX*.5+MIDTERMCS*.5
print("Midterm Grade is: ", "%.2f" % MIDTERMG)

FINALHOA=float(input("\55nEnter Hands-on Activity grade: "))
FINALQUIZ=float(input("Enter Final Quiz grade: "))
FINALAS=float(input("Enter Final Assignment grade: "))
FINALCS=FINALHOA*.5+FINALQUIZ*.3+FINALAS*.2
print("Final CS is: ", FINALCS)
FINALEX=float(input("Enter Final Exam grade: "))
FINALG=FINALEX*.5+FINALCS*.5
print("Final Grade is: ", "%.2f" % FINALG)

```
