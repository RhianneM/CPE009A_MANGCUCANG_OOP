```python
class Car:
  def __init__(self,name,mileage):
    self.name = name
    self.mileage = mileage
  def description(self):
    return f"The {self.name} car gives the mileage of {self.mileage}km/l"
car1 = Car("Toyota",3000)
car2 = Car("Honda",2000)
print(car1.description())
print(car2.description())
```

```python
class Animals:
    def __init__(self, name, habitat, temperature):
        self.name = name
        self.habitat = habitat
        self.temperature = temperature

    def feature(self):
        return f"The {self.name} which has a body temperature {self.temperature} lives in {self.habitat}"

mammals = Animals("Dog", "Land", "warm")
birds = Animals("Sparrow", "Air", "warm")
reptiles = Animals("snake", "terrestrial", "cold")

#print(mammals.name)
#print(mammals.habitat)
print(mammals.feature())
```

```python
#Create a class of Shapes (triangle, circle, square, rectangle), name properties (name and sides), method (function that compute for the area of each shape)

import math

class Shapes:
    def __init__(self, name, sides):
        self.name = name
        self.sides = sides
    def compute_area(self, measurement_1, measurement_2=0):
        if self.name == "Triangle":
            area = 0.5 * measurement_1 * measurement_2
        elif self.name == "Circle":
            area = math.pi * (measurement_1 ** 2)
        elif self.name == "Square":
            area = measurement_1 ** 2
        elif self.name == "Rectangle":
            area = measurement_1 * measurement_2
        else:
            return "Shape not recognized."
        return f"The {self.name} with {self.sides} sides has a total area of {area:.2f}"

triangle = Shapes("Triangle", 3)
circle = Shapes("Circle", 0)
square = Shapes("Square", 4)
rectangle = Shapes("Rectangle", 4)


print(triangle.compute_area(10, 5))
print(circle.compute_area(2))
print(square.compute_area(4))
print(rectangle.compute_area(8, 3))
```
