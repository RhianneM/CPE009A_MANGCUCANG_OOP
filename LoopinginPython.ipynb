**LOOPING IN PYTHON**


Loop Control Structures - used to perfrom looping operations until the given conditions is TRUE.

**TYPES OF LOOPS**

Loop Statement - A FOR loop is used for iteration over a sequence, similaar to while statements, but this statement is to peocess list in large range of numbers.

**LOOP STATEMENT**

**For Loop**

```python
#Ex 1:

week = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']
for x in week:
  print (x)
```

```python
#Ex 2

week = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']
for x in week:
  print (x)
  if x == 'Thursday':
    break
```

```python
#Ex 3

week = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']
for x in week:
  if x == 'Thursday':
    break
  print (x)
```

LOOPING THROUGH A STRING






The range () Function
- to loop through a set of code a specified number of times, we can use the range () function,
- The range () function returns a sequence of numbers, starting from 0 by default, and increments by 1
 (by default), and ends at a specified number.



```python
#Ex 4
  for n in range (6):
      print (n)

```

```python
for n in range (2,6):
      print (n)
```

```python
for n in range (2,6,2):
      print (n)
```

**NESTED LOOPS**

```python
#Ex 5
adjective = ['red', 'big', 'tasty']
fruits = ['apple', 'banana', 'cherry']

for x in adjective:
    for y in fruits:
      print (x,y)
```

**WHILE STATEMENT**

```python
#Ex 6

i = 1
while i < 6:
  print (i)
  i += 1
#i = i + 1
```

**THE BREAK STATEMENT**

```python
#Ex 7

i = 1
while i < 6:
  print (i)
  if i == 3:
    break
  i += 1
```

**THE CONTINUE STATEMENT**

```python
#Ex 8

i = 0
while i < 6:
  i += 1
  if i == 3:
   print (i)
  continue
```

**The Else Statement**

```python
#Ex 9

i = 1
while i < 6:
  print (i)
  i += 1
else:
  print ("i is no longer less than 6")
```
