**If Statement**

```python
a=12
b=100
if b>a:
  print ("b is greater than a")
```

**Elif Statements**

```python
a=12
b=12
if b>a:
  print("b is greater than a")
elif a==b:
  print("a and b are equal")
```

**Else Statememt**

```python
a=200
b=33
if b>a:
  print("b is greater than a")
elif a==b:
  print("a and b are equal")
else:
  print("a is greater than b")
```

**Short Hand If Statement**

```python
if a>b:
  print ("a is greater than b")
```

**Short Hand if..Else Statement**


```python
a=2
b=330
print ("A") if a>b else print ("B")
```

**And-both conditions are true**

```python
a=200
b=33
c=500
if a>b and c>a:
  print("Both conditions are true")
```

**Or-at leat one of the conditions is true**

```python
a=200
b=33
c=500
if a>b or a>c:
  print("At least one of the conditions is True")
```

**Nested if**

```python
x=41

if x>10:
  print ("Above ten,")
  else:
    print ("below 10.")
  if x>20:
    print ("and also above 20!")
  else:
    print ("below 20.")
  if x>30:
    print ("and also above 30!")
  else:
    print ("below 30.")
  if x>40:
    print ("and also above 40!")
  else:
    print("below 40")
  if x>50:
   print ("and also above 50!")
  else:
    print ("below 50.")
```
