```python

Calculate total marks and percentage
Assign grade for each subject based on marks:
90-100: A+

80-89: A

70-79: B+

60-69: B

50-59: C

40-49: D

Below 40: F (Fail)

Determine overall result (Pass/Fail)


Performance Remarks:

Percentage ≥ 90: "Outstanding"

Percentage ≥ 80: "Excellent"

Percentage ≥ 70: "Very Good"

Percentage ≥ 60: "Good"

Percentage ≥ 40: "Satisfactory"

Below 40: "Needs Improvement"
```

```python
#Calculate total marks and percentage:
subject = input("Enter subject: ")
marks = int(input("Enter marks: "))
total_marks = 100
Percentage = (marks/total_marks)*100
print ("Percentage: " , Percentage)
#Assign grade for each subject based on marks:
if marks >= 90:
    print ("A+")
elif marks >= 80:
    print ("A")
elif marks >= 70:
    print ("B+")
elif marks >= 60:
    print ("B")
elif marks >= 50:
    print ("C")
elif marks >= 40:
    print ("D")
else:
    print ("F")

#Performance Remarks:
if Percentage >= 90:
  print ("Outstanding")
elif Percentage >= 80:
  print ("Excellent")
elif Percentage >= 70:
  print ("Very Good")
elif Percentage >= 60:
  print ("Good")
elif Percentage >= 40:
  print ("Satisfactory")
else:
  print ("Needs Improvement")

#Determine overall result (Pass/Fail)
if marks >= 40:
  print ("Pass")
else:
  print ("Fail")

```
